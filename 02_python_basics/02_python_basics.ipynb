## 3.1 print() and comments
`print()` shows output. `#` starts a comment (Python ignores it).

```python
print("Hello, World!")
print("My name is", "Asha", "and I am learning", "Python")   # commas add spaces
print("A", "B", "C", sep="-")          # sep = separator
print("Loading", end="...")            # end = what comes at the end (default is new line)
print("done")

# f-strings (best way to mix text and variables)
name, age = "Asha", 19
print(f"{name} is {age} years old")
print(f"Next year: {age + 1}")
print(f"Pi rounded: {3.14159:.2f}")    # 2 decimal places
print("Line1\nLine2\tTabbed")           # \n new line, \t tab
"""This is a multi-line
comment / docstring"""
```

#### 🔍 How it works
- `print("A", "B")` → items separated by a space by default; `sep=` changes the separator.
- `end=` changes what is printed at the end (default is a new line).
- **f-string** `f"{x}"` → Python replaces `{x}` with the value; `:.2f` = 2 decimals; `:,` = thousands comma.
- `\n` new line, `\t` tab. `"""..."""` can hold text over many lines.
#### 💼 Where it's used
Printing is how you **debug** and show results. In jobs, `print` evolves into the `logging` module for production code, and f-strings are used for reports, file names and SQL/log messages.

```python
# Three ways to format output (all give the same result) — know all 3, interviews ask!
n1, n2 = 20, 30
total = n1 + n2
print('The addition of', n1, 'and', n2, 'is =', total)         # 1) commas
print('The addition of {} and {} is = {}'.format(n1, n2, total)) # 2) .format() with {} placeholders
print('The addition of {1} and {0} is = {2}'.format(n1, n2, total))  # position numbers can reorder
print(f'The addition of {n1} and {n2} is = {total}')              # 3) f-string (modern, fastest)
print('Score: %d, Average: %.2f' % (90, 87.456))                   # 4) old %-style
```

## 3.2 Variables
A variable is a **name that points to a value** (like a labelled box).

**Naming rules**
- Start with a letter or `_`, never a number.
- Only letters, numbers, `_`. No spaces or special characters.
- Case-sensitive (`age` ≠ `Age`). Can't use keywords like `if`, `for`, `class`.
- Style: `snake_case` (e.g. `student_name`).

```python
x = 10
name = "Asha"
price = 99.5
is_student = True
print(x, name, price, is_student)

a = b = c = 5                 # same value to many
p, q, r = 1, 2, 3             # many values at once
p, q = q, p                   # swap without temp!
print(a, b, c, p, q, r)

import keyword
print(keyword.kwlist)         # words you cannot use as variable names
```

#### 🔍 How it works
- `a, b, c = 1, 2, 3` unpacks several values at once; `p, q = q, p` swaps with no temp variable (Python builds a tuple first).
- `keyword.kwlist` lists the ~35 reserved words you can't use as names.
#### 💼 Where it's used
Clear variable names (`customer_age`, not `x1`) are the first thing code reviewers check. Swapping/unpacking appears in sorting algorithms and data loops.

### 3.2b Variables — what really happens in memory
A variable is a **name tag stuck on an object** in memory. `x = 3` doesn't put 3 *inside* x; it makes `x` point to the object `3`.
- `id(obj)` → the memory address of the object. `is` → checks "same object?"; `==` → checks "same value?".
- Python **re-uses small integers** (−5 to 256), so two names with value `30` point to the *same* object.
- **Identifier (variable name) rules:** letters, digits, `_` only · cannot start with a digit · case-sensitive · no spaces or symbols like `@ % $` · cannot be a keyword (`if`, `for`, `class`...).

```python
x = 3
y = 3
print(id(x) == id(y), x is y)          # True True  -> same object (small int caching)
x = 4                                  # x now points to a different object; y still points to 3
print(x, y, id(x) == id(y))

a = b = c = 30                         # chained assignment: all 3 names -> one object
print(a is b, id(b) == id(c))

big1, big2 = 1000, 1000
print("Equal values:", big1 == big2)   # True: values are equal
# 'is' on large numbers may be False in some contexts -> never use 'is' to compare values, use ==

# Invalid names: we run them with exec() inside try/except so the notebook doesn't stop
for bad in ["5 = v", "1nit = 27", "nit@ = 56", "nit% = 6", "for = 67"]:
    try:
        exec(bad)
    except SyntaxError as e:
        print(f"❌ {bad!r:12} -> SyntaxError: {e.msg}")

nit1, Nit1, _nit, nit_1 = 1, 2, 3, 4   # valid names (case-sensitive: nit1 and Nit1 are different)
print(nit1, Nit1)
try:
    print(NIT1)                         # NameError: Python is case-sensitive
except NameError as e:
    print("❌ NameError:", e)
```

#### 🔍 How it works
- `x is y` → `True` because both names point to the cached object `3`.
- Re-assigning `x = 4` does **not** change `y`: names are just labels.
- `exec("5 = v")` runs a string as code; the `try/except SyntaxError` catches the mistake and prints the reason. This shows *why* each name is illegal without crashing.
#### 💼 Where it's used
Understanding references prevents classic bugs (e.g. two lists accidentally sharing data — see dictionary/list copy section) and is a **very common interview topic** ("what does `is` vs `==` do?"). Good naming rules = readable, reviewable code.

## 3.4 Type casting (converting types)
- `int()`, `float()`, `str()`, `bool()`, `list()`, `tuple()`, `set()`
- `bool()` is `False` only for: `0`, `0.0`, `""`, `[]`, `()`, `{}`, `None`.

```python
print(int("25") + 5)          # str -> int
print(int(3.99))              # float -> int (cuts decimals, no rounding!)
print(float("2.5"))
print(str(100) + " rupees")
print(bool(0), bool(""), bool("False"), bool([0]))
print(list("abc"), tuple([1, 2]), set([1, 1, 2]))
print(round(3.567, 2), round(2.5), round(3.5))   # banker's rounding: .5 goes to even

# What if conversion fails?
try:
    int("hello")
except ValueError as e:
    print("Error:", e)
```

### 3.4b Numbers deep-dive: bool, complex & casting every type
- `bool` is a subclass of `int`: `True == 1`, `False == 0`, so `True + True = 2`.
- **complex** numbers: `a + bj` with `.real` and `.imag` (used in signal processing, physics, FFT).

```python
print(True + True, True * 10, False // True, 7 // True)     # bool behaves as 1/0 in maths
c, c1 = 10 + 20j, 20 + 30J
print(type(c), c.real, c.imag, c + c1, c * c1)

print(int(True), float(False), complex(True), complex('10'), complex(10, 20))
print(str(10), str(10.4), str(True), str(1 + 2j))
for bad in ["int('ten')", "int('10.5')", "complex('ten')", "float('abc')", "int([1, 2])"]:
    try: eval(bad)
    except (ValueError, TypeError) as e: print(f"❌ {bad:16} -> {type(e).__name__}: {e}")
print(int(float('10.5')), bin(int('1010', 2)))              # two-step conversion works
```

#### 🔍 How it works
- `False // True` → `0 // 1 = 0`. Floor division `//` drops the decimals (rounds down).
- `c.real`, `c.imag` read the two parts of a complex number. Python writes it as `(10+20j)`.
- `int('10.5')` fails because a string with a decimal point is not an integer text; do `int(float('10.5'))`.
- `eval("...")` executes a string as an expression – we only use it here on safe literals.
#### 💼 Where it's used
**Type casting is everyday data cleaning**: CSV columns arrive as text and must become numbers (`pd.to_numeric`, `astype`). Failing casts (`ValueError`) are the #1 data-loading error.

## 3.6 Operators
| Group | Operators |
|---|---|
| Arithmetic | `+ - * / // % **` |
| Comparison | `== != > < >= <=` |
| Logical | `and or not` |
| Assignment | `= += -= *= /= //= %= **=` |
| Membership | `in`, `not in` |
| Identity | `is`, `is not` |

```python
a, b = 17, 5
print(a + b, a - b, a * b)
print(a / b)     # true division -> float
print(a // b)    # floor division
print(a % b)     # remainder
print(a ** b)    # power

print(a > b, a == b, a != b)
print(a > 10 and b < 10, a > 20 or b < 10, not (a > b))

n = 10
n += 5; n *= 2; n -= 3
print(n)

print("a" in "apple", 3 not in [1, 2, 3])
l1, l2 = [1, 2], [1, 2]
print(l1 == l2, l1 is l2)      # same value vs same object
print(0.1 + 0.2 == 0.3)        # False! floating point precision
import math; print(math.isclose(0.1 + 0.2, 0.3))
```

#### 🔍 How it works
- `/` always gives a float (`17/5 = 3.4`); `//` floor-divides (3); `%` gives the remainder (2); `**` is power.
- `and / or / not` combine True/False conditions; `in` tests membership; `is` tests identity.
- `0.1 + 0.2 == 0.3` is `False` because computers store decimals in binary (tiny rounding error) → compare with `math.isclose`.
#### 💼 Where it's used
`%` finds even/odd, cycles (days of week), and batching; `//` converts seconds to minutes; the float issue matters in **finance** (use `decimal` for money).

## 3.7 Number systems & Bitwise operators
Computers store everything in **binary (base 2)**.
| System | Base | Digits | Python |
|---|---|---|---|
| Binary | 2 | 0-1 | `0b1010` |
| Octal | 8 | 0-7 | `0o12` |
| Decimal | 10 | 0-9 | `10` |
| Hexadecimal | 16 | 0-9, A-F | `0xA` |

**Bitwise operators work on the bits:**
| Op | Name | Example (5=0101, 3=0011) |
|---|---|---|
| `&` | AND | `5 & 3 = 1` |
| `\|` | OR | `5 \| 3 = 7` |
| `^` | XOR | `5 ^ 3 = 6` |
| `~` | NOT | `~5 = -6` |
| `<<` | left shift (×2) | `5 << 1 = 10` |
| `>>` | right shift (÷2) | `5 >> 1 = 2` |

```python
n = 10
print(bin(n), oct(n), hex(n))            # decimal -> other systems
print(int("1010", 2), int("12", 8), int("A", 16))   # other -> decimal
print(0b1010, 0o12, 0xA)

a, b = 5, 3
print(f"{a:04b} & {b:04b} = {a & b:04b}")
print(a | b, a ^ b, ~a, a << 1, a >> 1)

# Trick: even/odd using bitwise
for num in [4, 7]:
    print(num, "odd" if num & 1 else "even")
```

#### 🔍 Line-by-line
- `bin(n), oct(n), hex(n)` → text in base 2 / 8 / 16 (`0b1010`, `0o12`, `0xa`).
- `int("1010", 2)` → the second number says "this text is in base 2".
- `5 & 3` → bitwise AND: compare bits `0101` and `0011` → `0001` = 1. `<<` shifts bits left = multiply by 2.
- `num & 1` → looks only at the last bit: 1 = odd, 0 = even (a trick faster than `%`).
#### 💼 Where it's used
Bitwise operations power **permissions/flags**, networking (IP masks), compression, hashing and low-level optimisation. Binary/hex knowledge is needed for colour codes (`#FF5733`) and memory/debugging.

## 🚀 Mini Project 1: Smart Bill Splitter
Uses: variables, input, arithmetic, f-strings, casting.

```python
def bill_splitter(total, tip_percent, people):
    tip = total * tip_percent / 100
    grand_total = total + tip
    per_person = grand_total / people
    print("=" * 30)
    print(f"Bill: ₹{total:.2f}")
    print(f"Tip ({tip_percent}%): ₹{tip:.2f}")
    print(f"Grand total: ₹{grand_total:.2f}")
    print(f"Each of {people} pays: ₹{per_person:.2f}")
    print("=" * 30)

bill_splitter(1250, 10, 4)
# Interactive version:
# bill_splitter(float(input("Bill: ")), float(input("Tip %: ")), int(input("People: ")))
```

#### 🔍 How it works
- A **function** `bill_splitter(total, tip_percent, people)` packs the steps; we call it with different numbers without rewriting the maths.
- `tip = total * tip_percent / 100`, `grand_total = total + tip`, `per_person = grand_total / people`; `{value:.2f}` formats money to 2 decimals; `"=" * 30` draws a line.
#### 💼 Where it's used
Every calculator in billing, e-commerce checkout and fintech is this: **inputs → formula → formatted output**.
