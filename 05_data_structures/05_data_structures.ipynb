## 6.1 List — every method

```python
fruits = ["apple", "banana", "cherry"]
print(fruits[0], fruits[-1], fruits[0:2], len(fruits))

fruits.append("mango")               # add at end
fruits.insert(1, "kiwi")             # add at position
fruits.extend(["grape", "lemon"])    # add many
print(fruits)

fruits.remove("banana")              # remove by value
last = fruits.pop()                  # remove last (returns it)
second = fruits.pop(1)               # remove by index
del fruits[0]                        # delete by index
print(fruits, last, second)

nums = [5, 2, 9, 1, 5, 6]
nums.sort(); print(nums)                    # ascending, in place
nums.sort(reverse=True); print(nums)
print(sorted(nums), nums)                   # sorted() returns new list
nums.reverse(); print(nums)
print(nums.count(5), nums.index(9))
print(max(nums), min(nums), sum(nums), len(nums))

copy1 = nums.copy()                  # real copy (nums2 = nums would NOT copy)
nums.clear(); print(nums, copy1)

# Slicing, concatenation, repetition, membership, nested
a = [1, 2, 3]; b = [4, 5]
print(a + b, a * 2, 3 in a)
matrix = [[1, 2, 3], [4, 5, 6]]
print(matrix[1][2])
# Unpacking
first, *middle, last = [1, 2, 3, 4, 5]
print(first, middle, last)
```

#### 🔍 How it works (lists)
- `append(x)` adds one item at the end; `insert(i, x)` at position i; `extend([..])` adds many.
- `remove(x)` deletes by **value**; `pop(i)` deletes by **index** and returns it; `del l[i]` deletes by index.
- `sort()` changes the list in place and returns `None`; `sorted(l)` returns a **new** list.
- `nums2 = nums` makes a *second name*; `nums.copy()` makes a real copy.
- `first, *middle, last = [...]` → star-unpacking grabs "everything in between".
#### 💼 Where it's used
Lists are the most used structure: rows of data, API results, queue of tasks. Every DataFrame column is conceptually a list.

### 6.1b List: more class examples
- `l.append(20, 30, 40)` ❌ fails: `append` takes **one** item. Use `extend([20,30,40])` or `+=`.
- Lists can hold **mixed types and other lists** (nested). Strings inside lists are indexed too: `l1[4][0]`.
- Negative step slicing `[::-2]` walks backward, skipping one.

```python
l = []
l.append(10)
try:
    l.append(20, 30, 40)
except TypeError as e:
    print("❌", e)
l.extend([20, 30, 40]); print(l)

l1 = [10, 3.5, True, 1 + 2j, 'welcome']
l1.append([1, 2, 3])                          # a list inside a list (nested)
print(l1, len(l1))
print(l1[4], l1[4][0], l1[4][-1], l1[5][1])   # index into the string / nested list
print(l1[::-2])                               # backward, every 2nd element
l1[0] = 100                                   # lists are mutable
print(l1)
try:
    print(l1[4][7])                           # 'welcome' has indexes 0..6
except IndexError as e:
    print("❌ IndexError:", e)
```

## 6.2 Tuple — like a list that cannot change
Use for fixed data (coordinates, days of week, DB rows). Faster and safer than lists.

```python
t = (10, 20, 30, 20)
print(t[0], t[-1], t[1:3], len(t))
print(t.count(20), t.index(30))
single = (5,)                # comma makes it a tuple
print(type(single), type((5)))
x, y, z, w = t               # unpacking
print(x, y)
# t[0] = 99                  # TypeError: tuples are immutable
print(tuple(sorted(t)), t + (1, 2))
```

#### 🔍 How it works / 💼 Where it's used
- A tuple is a **fixed** sequence — safe from accidental change and faster than a list. `(5,)` needs the comma, otherwise `(5)` is just the number 5.
- Unpacking `x, y, z, w = t` assigns many variables at once. Functions that return several values return a **tuple**.
- Used for coordinates `(lat, lon)`, database rows, dictionary keys (lists can't be keys, tuples can), and constants.

## 6.3 Set — unique items, fast lookup, maths operations

```python
s = {1, 2, 3, 3, 2}
print(s)                       # duplicates removed
s.add(10); s.update([20, 30])
s.remove(1)                    # error if missing
s.discard(999)                 # no error if missing
print(s, s.pop())

A, B = {1, 2, 3, 4}, {3, 4, 5, 6}
print("Union:", A | B, A.union(B))
print("Intersection:", A & B, A.intersection(B))
print("Difference:", A - B, A.difference(B))
print("Symmetric diff:", A ^ B)
print({1, 2}.issubset(A), A.issuperset({1}), A.isdisjoint({9}))
print(list(set([1, 1, 2, 3, 3])))   # quickest way to remove duplicates
```

#### 🔍 How it works (sets)
- Sets store **unique** items with no order, so no indexing. `remove()` errors if missing; `discard()` doesn't.
- `A | B` union, `A & B` intersection, `A - B` only in A, `A ^ B` in exactly one → the same maths as Venn diagrams.
#### 💼 Where it's used
Unique users, "which customers bought X but not Y", finding new vs churned customers, tag matching — set logic runs fast even on millions of items.

### 6.3b Frozenset (an unchangeable set), subset / superset / disjoint
`frozenset` = a set that **cannot be changed** → can be used as a dictionary key or inside another set.

```python
s1 = {10, 20, 30}
fs = frozenset(s1)
print(type(fs), fs)
try:
    fs.add(60)
except AttributeError as e:
    print("❌", e)
A, B, C = {1, 2, 3, 4}, {1, 2}, {8, 9}
print("B subset of A:", B <= A, "| A superset of B:", A >= B, "| A and C disjoint:", A.isdisjoint(C))
print({frozenset({1, 2}): "pair"})            # frozenset as a dict key (a normal set would fail)
print("A - B:", A - B, "| sorted(A | C):", sorted(A | C))
```

#### 💼 Where it's used
Sets give **O(1) membership tests** — checking `x in big_set` is far faster than `x in big_list`. Used for de-duplication (unique customers), blacklists, tag matching, and comparing two datasets (what is new/missing).

## 6.4 Dictionary — key : value pairs
Perfect for structured data (like JSON). Keys must be unique & immutable.

```python
student = {"name": "Asha", "age": 19, "course": "Data Science"}
print(student["name"], student.get("age"), student.get("city", "N/A"))

student["city"] = "Delhi"            # add
student["age"] = 20                      # update
student.update({"year": 2, "age": 21})   # add/update many
print(student)

print(student.keys(), student.values(), student.items())
for k, v in student.items():
    print(f"{k} -> {v}")

removed = student.pop("year")            # remove a key
student.popitem()                        # remove last inserted
print(student, removed)
print("name" in student)                 # checks keys
student.setdefault("skills", []).append("Python")
print(student)
print(dict.fromkeys(["a", "b"], 0))

# Nested dict
data = {"Riya": {"maths": 90, "sci": 85}, "Arjun": {"maths": 70, "sci": 95}}
print(data["Riya"]["sci"])
# Dictionary comprehension
squares = {n: n**2 for n in range(1, 6)}
print(squares)
# Sorting by value
scores = {"a": 50, "b": 90, "c": 70}
print(sorted(scores.items(), key=lambda kv: kv[1], reverse=True))
# Merge (Python 3.9+)
print({"x": 1} | {"y": 2})
# copy = safe duplicate; copy.deepcopy for nested
import copy
deep = copy.deepcopy(data)
```

#### 🔍 How it works (dictionary)
- `d[key]` raises `KeyError` if missing; **`d.get(key, default)`** doesn't → prefer `get` for safe lookups.
- `keys()`, `values()`, `items()` give views; `for k, v in d.items()` loops through pairs.
- `pop(key)` removes a key; `setdefault(key, default)` creates it if missing; `update({...})` merges.
- Dict comprehension `{n: n**2 for n in range(1, 6)}` builds a lookup in one line; `sorted(d.items(), key=lambda kv: kv[1])` sorts by value.
#### 💼 Where it's used
**JSON / API data is dictionaries.** Config files, counting (`word → count`), lookup tables, caching, and converting to DataFrames (`pd.DataFrame(list_of_dicts)`).

### 6.4b Copy vs reference (the most common list/dict bug)
`d2 = d1` does **not** copy — both names point to the same dict. Change one → the other changes. Use `.copy()` (shallow) or `copy.deepcopy()` (fully independent, for nested data).

```python
d1 = {"a": 1, "b": [10, 20]}
d2 = d1                       # same object
d3 = d1.copy()                # shallow copy: new dict, but inner list is still shared
d4 = copy.deepcopy(d1)        # deep copy: everything new
print(d1 is d2, d1 is d3, d1 is d4, id(d1) == id(d2))

d1["a"] = 99
d1["b"].append(30)
print("d2 (reference) :", d2)       # sees both changes
print("d3 (shallow)   :", d3)       # sees only the inner-list change
print("d4 (deep)      :", d4)       # unaffected

r = range(2, 20, 4)           # range is an immutable sequence type (stores only start, stop, step)
print(list(r), len(r), 10 in r, r[2], type(r))
```

#### 🔍 How it works
- `is` compares *identity* (same object?). `d3 = d1.copy()` creates a new dict, but `d1["b"]` (a list) is still the *same list* in both → appending shows in `d3`.
- `range(2, 20, 4)` doesn't build a list; it just remembers the rule → memory-efficient even for `range(10**9)`.
#### 💼 Where it's used
Accidental shared references cause hard-to-find bugs in **data pipelines and ML code** (modifying a "copy" of a DataFrame that was really a view — pandas `SettingWithCopyWarning`). Use `df.copy()` before editing.

## 🚀 Mini Project 4: Contact Book + Word Counter + Unique Visitors

```python
# --- Contact book (dictionary) ---
contacts = {}
def add(name, phone): contacts[name] = phone
def find(name): return contacts.get(name, "Not found")
def delete(name): return contacts.pop(name, "Nothing to delete")
add("Riya", "98765"); add("Arjun", "91234")
print(find("Riya"), find("Zed"), delete("Arjun"), contacts)

# --- Word frequency (dict) ---
text = "the cat and the hat and the bat"
freq = {}
for w in text.split():
    freq[w] = freq.get(w, 0) + 1
print(freq)
print(max(freq, key=freq.get))          # most common word
from collections import Counter
print(Counter(text.split()).most_common(2))

# --- Unique visitors (set) ---
day1, day2 = {"a", "b", "c"}, {"b", "c", "d"}
print("Returning:", day1 & day2, "| New on day 2:", day2 - day1, "| Total unique:", len(day1 | day2))

# --- To-do list (list) ---
todo = ["study", "gym"]; todo.append("code"); todo.remove("gym")
for i, task in enumerate(todo, 1): print(f"{i}. {task}")
```

#### 🔍 How it works
- **Contact book:** a dict maps name → phone; functions wrap add / find / delete (`pop(name, default)` avoids errors).
- **Word frequency:** `freq.get(w, 0) + 1` counts without needing `if w in freq`. `Counter(...).most_common(n)` does it in one line. `max(freq, key=freq.get)` returns the most common word.
- **Unique visitors:** set maths (`&`, `-`, `|`) answers "returning / new / total".
#### 💼 Where it's used
Counting and grouping is the heart of analytics: top products, most frequent errors in logs, NLP word counts.
