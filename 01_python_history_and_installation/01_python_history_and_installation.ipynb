# 1️⃣ Python History & Why Python
### 🟢 What is Python?
Python is a **high-level, interpreted, general-purpose** programming language. Code reads almost like English, so beginners learn fast, and professionals use it for web, data science, AI, automation and more.

### History in short
- **1989 (Dec):** Guido van Rossum starts Python as a hobby project at CWI, Netherlands.
- **1991 (Feb):** Python 0.9.0 released.
- **The name** comes from the comedy show *Monty Python's Flying Circus*, not the snake.
- **2000:** Python 2.0 (list comprehensions, garbage collection).
- **2008:** Python 3.0 (cleaner, not backward compatible).
- **2020 (Jan 1):** Python 2 officially retired.
- **Today:** Python 3.x, one new version every October.

### Why Python for Data Science & AI?
- Easy syntax → focus on ideas, not semicolons.
- Huge libraries: NumPy, Pandas, Matplotlib, Seaborn, Scikit-learn, TensorFlow, PyTorch.
- Massive community, free, works on Windows/Mac/Linux.

````python
import matplotlib.pyplot as plt

events = [(1991, "Python 0.9\nreleased"), (1994, "Python 1.0"), (2000, "Python 2.0"),
          (2008, "Python 3.0"), (2020, "Python 2\nretired"), (2024, "Python 3.13")]
years = [e[0] for e in events]

fig, ax = plt.subplots(figsize=(11, 3.2))
ax.hlines(0, 1988, 2026, color="gray")
for i, (y, label) in enumerate(events):
    ax.plot(y, 0, "o", color="#3776ab", markersize=10)
    ax.text(y, 0.35 if i % 2 == 0 else -0.55, f"{y}\n{label}", ha="center", fontsize=9)
ax.set_ylim(-1, 1); ax.axis("off"); ax.set_title("Python Timeline", fontsize=14)
plt.show()
````

#### 🔍 How the chart is built
- `plt.subplots(figsize=(11, 3.2))` → creates the canvas (`fig`) and one drawing area (`ax`); `figsize` is width × height in inches.
- `ax.hlines(0, 1988, 2026)` draws the horizontal time line; `ax.plot(year, 0, "o")` puts a dot per event; `ax.text(...)` writes the label above/below (alternating with `i % 2` so labels don't overlap).
- `ax.axis("off")` hides the axes so it looks like a clean timeline.
#### 💼 Where it's used
Timelines/roadmaps in reports. The skill reused everywhere: **create figure → draw layers → label → show**.

### 1.1 How Python runs your code — interpreter, compiler, PVM
| | **Interpreter** | **Compiler** |
|---|---|---|
| Translates | one statement at a time | the whole program at once |
| On error | stops at that line (easy to debug) | no output until the whole program is fixed |
| Examples | Python, JavaScript | C, C++, Java (to bytecode) |

- Python is **interpreted**, but internally: `your .py code → bytecode (.pyc) → Python Virtual Machine (PVM) → machine instructions`.
- **PVM** is why the *same code runs on Windows, Mac and Linux* (platform independent).
- **IDE** (Integrated Development Environment) = editor + run + debug in one tool: VS Code, PyCharm, Spyder, Jupyter.

### 1.4 Facts worth remembering (also interview favourites)
- Created by **Guido van Rossum** (Netherlands); first release 1991; named after *Monty Python's Flying Circus*.
- Features: easy syntax, dynamically typed (no need to declare types), object-oriented, open source, huge standard library, extensible, portable.
- Used by Google, NASA, Netflix, Uber, Instagram, Spotify, Reddit.
- **Weak spots:** not used for native mobile apps; slower than C/C++ for raw speed (so heavy maths is done by NumPy/C libraries underneath).
- **Fix "python is not recognized" (Windows):** search *Environment Variables → Path → New* and add your Python folder and its `Scripts` folder, then reopen cmd.

### 2.1 Install Python (python.org)
1. Go to **python.org → Downloads** and download the latest Python 3.
2. Run the installer. ✅ **Tick "Add Python to PATH"** (most common beginner mistake!).
3. Click *Install Now*.
4. Open Command Prompt and check:
```
python --version
pip --version
```

### 2.2 Install Anaconda (recommended for Data Science)
Anaconda = Python + Jupyter + 250+ data science libraries + `conda` package manager.
1. Download from **anaconda.com/download** (choose Windows/Mac/Linux).
2. Run the installer → *Just Me* → keep default location.
3. Open **Anaconda Prompt** and check: `conda --version`
4. Launch with: `jupyter notebook` or `jupyter lab`

![Anaconda installer](images/02_anaconda_installer.png)

### 2.3 Install VS Code
1. Download from **code.visualstudio.com**.
2. Install extensions: **Python** and **Jupyter** (both by Microsoft).
3. Press `Ctrl+Shift+P` → *Python: Select Interpreter* → choose your Anaconda / Python.
4. Create a file `hello.py` and press ▶ Run.

### 2.4 pip vs conda
| Task | pip | conda |
|---|---|---|
| Install | `pip install pandas` | `conda install pandas` |
| Upgrade | `pip install -U pandas` | `conda update pandas` |
| Uninstall | `pip uninstall pandas` | `conda remove pandas` |
| List | `pip list` | `conda list` |

### 2.5 🔵 Virtual environments (professional habit)
Each project gets its own clean set of libraries.
```
python -m venv myenv
myenv\Scripts\activate        # Windows
source myenv/bin/activate     # Mac/Linux
pip install pandas numpy
pip freeze > requirements.txt
```
With conda: `conda create -n ds python=3.12` then `conda activate ds`.

````python
import os
from IPython.display import Image, display

def show_image(name, width=600):
    """Show an image from the images/ folder if it exists."""
    path = os.path.join("images", name)
    if os.path.exists(path):
        display(Image(path, width=width))
    else:
        print(f"(add your screenshot at {path})")

show_image("01_python_installer.png")

# Check your setup
import sys, platform
print("Python version:", sys.version.split()[0])
print("OS:", platform.system())
````
