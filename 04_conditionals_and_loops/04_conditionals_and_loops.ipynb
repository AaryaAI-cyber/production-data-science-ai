## 5.1 if / elif / else
Python uses **indentation (4 spaces)** instead of braces.

```python
marks = 78
if marks >= 90:
    grade = "A+"
elif marks >= 75:
    grade = "A"
elif marks >= 60:
    grade = "B"
else:
    grade = "Fail/C"
print("Grade:", grade)

# One-line (ternary)
print("Pass" if marks >= 40 else "Fail")

# Nested + logical operators
age, has_id = 20, True
if age >= 18 and has_id:
    print("Entry allowed")

# Truthy / Falsy
items = []
if not items:
    print("List is empty")

# match-case (Python 3.10+)
day = 3
match day:
    case 1: print("Mon")
    case 2: print("Tue")
    case 3: print("Wed")
    case _: print("Other")
```

### 5.1b Nested if, and why `if / elif` differs from many separate `if`s
- **Separate `if`s** → *every* condition is checked, even after one matches.
- **`if / elif / else`** → Python stops at the **first** True branch (faster and avoids double output).
- **Nested if** → an `if` inside an `if` (second check happens only if the first is True).

```python
x = 4
r = x % 2                                   # remainder when dividing by 2
if r == 0:
    print('Even number')
    if x > 5:                               # nested: runs only for even numbers
        print('greater than 5')
    else:
        print('5 or less')
else:
    print('Odd number')

n = 1
print("--- separate ifs (all checked) ---")
if n == 1: print('one')
if n == 1 or n == 2: print('one or two')    # also prints!
print("--- elif (stops at first match) ---")
if n == 1: print('one')
elif n == 1 or n == 2: print('one or two')  # skipped because first branch matched
else: print('number not found')

# Mistake to avoid: this always prints "odd number" even for even numbers
x = 4
if x % 2 == 0: print('Even number')
print('odd number')                         # not inside any if -> always runs
age = 19
if age > 18: print("Eligible to vote.")     # one-line if
```

#### 🔍 How it works
- `x % 2` is the remainder; remainder 0 means even. Python decides by **indentation** (4 spaces) — the line `print('odd number')` at the left edge isn't part of the `if`, so it always runs.
- With `elif`, once `n == 1` matched, the rest is skipped.
#### 💼 Where it's used
Business rules: discount tiers, eligibility, risk flags, data validation (`if value is None`), feature flags in apps. Rule engines in banks and insurance are chains of these.

## 5.2 for loop & range()
`range(stop)`, `range(start, stop)`, `range(start, stop, step)`.

```python
for i in range(5): print(i, end=" ")
print()
for i in range(2, 10, 2): print(i, end=" ")
print()
for i in range(10, 0, -3): print(i, end=" ")
print()
for ch in "Hi!": print(ch, end="|")
print()
for idx, fruit in enumerate(["apple", "mango", "kiwi"], start=1):
    print(idx, fruit)
for a, b in zip([1, 2, 3], ["x", "y", "z"]):
    print(a, b)
# for-else: else runs if no break
for n in range(3):
    pass
else:
    print("loop finished normally")
```

#### 🔍 How it works
- `range(2, 10, 2)` → 2, 4, 6, 8 (stop excluded). `range(10, 0, -3)` counts down.
- `enumerate(list, start=1)` gives `(index, item)` pairs; `zip(a, b)` walks two lists together.
- **for-else**: the `else` block runs only if the loop finished without `break`.
#### 💼 Where it's used
Looping over rows/files/API pages. (In Pandas/NumPy prefer *vectorised* operations; plain loops are slow on millions of rows.)

## 5.3 while loop, break, continue, pass
- `break` → stop the loop. `continue` → skip to next round. `pass` → do nothing (placeholder).

```python
count = 1
while count <= 5:
    print("count =", count)
    count += 1

for n in range(1, 11):
    if n == 3: continue      # skip 3
    if n == 7: break         # stop at 7
    print(n, end=" ")
print()

# Nested loops: multiplication table pattern
for i in range(1, 4):
    for j in range(1, 6):
        print(i * j, end="\t")
    print()

# Star pyramid
for i in range(1, 5):
    print(" " * (4 - i) + "*" * (2 * i - 1))
```

### 5.3b While-loop recipe (init → condition → update) + patterns
Every `while` loop has 3 parts: **initialise** → **condition** → **update**. Forget the update = infinite loop.

```python
i = 1                                        # 1) initialise
while i <= 5:                                # 2) condition
    print('data science', i)
    i += 1                                   # 3) update (increment)

i = 5
while i >= 1:                                # counting down
    print(i, end=' ')
    i -= 1
print()

# Nested while: for every i, inner loop runs fully
i = 1
while i <= 3:
    print('row', i, end=' : ')
    j = 1
    while j <= 4:
        print('tech', end=' - ')             # end=' - ' keeps output on one line
        j += 1
    i += 1
    print()                                  # new line after each row
```

#### Pattern programs (a classic interview round)
Rule: **outer loop = rows, inner loop = columns/characters in that row.**

```python
n = 5
print("Right triangle"); [print("* " * i) for i in range(1, n + 1)]
print("\nInverted");     [print("* " * i) for i in range(n, 0, -1)]
print("\nNumber pyramid")
for i in range(1, n + 1):
    print(" " * (n - i) + " ".join(str(k) for k in range(1, i + 1)))
print("\nDiamond")
for i in list(range(1, n + 1)) + list(range(n - 1, 0, -1)):
    print(" " * (n - i) + "* " * i)
print("\nFloyd's triangle")
k = 1
for i in range(1, 5):
    for j in range(i):
        print(k, end=" "); k += 1
    print()
```

#### 🔍 How it works
- `" " * (n - i)` repeats a space to push the stars to the right → forms a centred shape. `"* " * i` repeats the star i times, so row 3 has 3 stars.
- `[print(...) for i in range(...)]` is a one-line loop (list comprehension used only for its side effect).
#### 💼 Where it's used
Patterns train **loop thinking**, and the same idea (rows × columns) is how you build tables, reports, game boards, matrices and image pixel loops.

## 🚀 Mini Project 3: Number Guessing Game + FizzBuzz

```python
import random
secret = random.randint(1, 20)
guesses = [10, 15, 5, secret]     # simulated user guesses (replace with input())
for attempt, g in enumerate(guesses, 1):
    if g < secret: print(f"Try {attempt}: {g} is too low")
    elif g > secret: print(f"Try {attempt}: {g} is too high")
    else:
        print(f"Try {attempt}: {g} is correct! 🎉"); break

# FizzBuzz
for i in range(1, 16):
    print("FizzBuzz" if i % 15 == 0 else "Fizz" if i % 3 == 0 else "Buzz" if i % 5 == 0 else i, end=" ")
```

#### 🔍 How it works
- **Guessing game:** `random.randint(1, 20)` picks the secret; each guess is compared with `<`/`>`; `break` ends the loop when correct.
- **FizzBuzz:** checks `% 15` first (divisible by both 3 and 5) *before* `% 3` and `% 5` — order matters!
- Python's one-line `A if cond else B` is the *conditional expression*.
#### 💼 Where it's used
FizzBuzz is a screening question in many interviews; the "compare & give hint" pattern is used in search algorithms (binary search).
