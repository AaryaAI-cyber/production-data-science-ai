# 4️⃣ Strings
A string is a sequence of characters inside `' '`, `" "` or `""" """`. Strings are **immutable** (cannot be changed in place).
## 4.1 Indexing & slicing
```
 P  y  t  h  o  n
 0  1  2  3  4  5      (forward index)
-6 -5 -4 -3 -2 -1      (backward index)
```
**Slicing:** `s[start : stop : step]` — start included, stop **excluded**.

````python
s = "Python"
print(s[0], s[-1], s[2])
print(s[0:3])      # Pyt
print(s[:3])       # Pyt
print(s[3:])       # hon
print(s[:])        # copy
print(s[::2])      # Pto  (every 2nd)
print(s[::-1])     # nohtyP (reverse!)
print(s[-3:])      # hon
print(s[5:1:-1])   # negative step goes backward
print(s[10:20])    # out-of-range slicing is safe -> ''
# print(s[10])     # IndexError: single index out of range is NOT safe
````

#### 🔍 How slicing works
- Index starts at **0**; negative index counts from the end (`-1` = last).
- `s[start:stop:step]` → **start included, stop excluded**; omitted start = beginning, omitted stop = end.
- `s[::-1]` → step −1 reads backward → **reverse a string** (classic interview question).
- A single bad index (`s[10]`) raises `IndexError`, but an out-of-range *slice* returns `''` safely.
#### 💼 Where it's used
Extracting parts of text: area codes from phone numbers, year from a date string (`d[:4]`), file extensions (`name[-4:]`), parsing logs.

## 4.2 String methods
| Method | Use |
|---|---|
| `upper() lower() title() capitalize() swapcase()` | change case |
| `strip() lstrip() rstrip()` | remove spaces |
| `replace(a,b)` | replace text |
| `split(sep)` / `join(list)` | break / combine |
| `find() index() count()` | search |
| `startswith() endswith()` | check start/end |
| `isdigit() isalpha() isalnum() isspace() isupper() islower()` | check content |
| `center() ljust() rjust() zfill()` | padding |

````python
t = "  Hello, Data Science World!  "
print(t.strip())
print(t.upper(), "|", t.lower(), "|", t.title(), "|", t.swapcase())
print(t.replace("World", "Universe"))
words = t.strip().split(" ")
print(words)
print("-".join(words))
print(t.find("Data"), t.find("xyz"), t.count("l"))
print(t.strip().startswith("Hello"), t.strip().endswith("!"))
print("123".isdigit(), "abc".isalpha(), "abc123".isalnum(), " ".isspace())
print("7".zfill(3), "hi".center(10, "*"), "hi".ljust(6, ".") + "|", "hi".rjust(6, ".") )
print("a,b,,c".split(","), "one two  three".split())
print("Line1\nLine2".splitlines())
print(len(t), "ell" in t, t.strip()[::-1])
# format styles
name, score = "Riya", 92.456
print("Name: %s Score: %.1f" % (name, score))
print("Name: {} Score: {:.1f}".format(name, score))
print(f"Name: {name} Score: {score:.1f} | {score:>10.2f} | {1234567:,}")
````

#### 🔍 How it works
- Strings are **immutable**: methods like `upper()` or `replace()` **return a new string**; the original doesn't change.
- `strip()` removes spaces; `split(" ")` → list of words; `"-".join(list)` → glues them back; `find()` → position or −1; `count()` → how many times.
- `isdigit/isalpha/isalnum` validate input; `zfill(3)` pads with zeros (`007`); `{score:>10.2f}` right-aligns in 10 spaces with 2 decimals.
#### 💼 Where it's used
**Text cleaning** is half of data science: trimming spaces in names, lower-casing for matching, splitting "City, State", validating IDs. In Pandas the same methods are `df.col.str.strip()`, `.str.lower()`, `.str.split()`.

## 🚀 Mini Project 2: Password Strength Checker
Uses: strings, methods, conditionals-like logic.

````python
def check_password(pw):
    score = 0
    tips = []
    if len(pw) >= 8: score += 1
    else: tips.append("Use at least 8 characters")
    if any(c.isupper() for c in pw): score += 1
    else: tips.append("Add an uppercase letter")
    if any(c.islower() for c in pw): score += 1
    else: tips.append("Add a lowercase letter")
    if any(c.isdigit() for c in pw): score += 1
    else: tips.append("Add a number")
    if any(c in "!@#$%^&*" for c in pw): score += 1
    else: tips.append("Add a symbol like @ or #")
    level = ["Very weak", "Weak", "Okay", "Good", "Strong", "Excellent"][score]
    return level, tips

for p in ["abc", "Python123", "P@ssw0rd!Long"]:
    print(p, "->", check_password(p))
````

#### 🔍 How it works
- Each rule uses `any(c.isupper() for c in pw)` → a **generator** that stops as soon as one character matches (efficient).
- A `score` counter + a `tips` list collect results; the score picks a label from a list by index.
- Function returns **two values** (level, tips) as a tuple.
#### 💼 Where it's used
Input validation on every sign-up form; the same "check rules → give feedback" design is used in data quality checks (e.g. "is the email valid? is age between 0 and 120?").
