## 8.1 Exception handling

```python
try:
    num = int("abc")
    result = 10 / 0
except ValueError:
    print("Not a number")
except ZeroDivisionError:
    print("Cannot divide by zero")
except Exception as e:
    print("Other error:", e)
else:
    print("No error")
finally:
    print("Always runs")

def age_check(age):
    if age < 0:
        raise ValueError("Age cannot be negative")
    return age
try: age_check(-1)
except ValueError as e: print("Caught:", e)
```

#### 🔍 How it works
- `try` runs risky code; `except ValueError` runs if that error type happens; `else` runs if **no** error; `finally` **always** runs (closing files/connections).
- `raise ValueError("...")` throws your own error to stop invalid data early.
#### 💼 Where it's used
Production code never trusts input: reading files that may not exist, API calls that time out, bad rows in CSVs. Good error handling = programs that don't crash at 3 am.

## 8.2 File handling & JSON
Modes: `r` read, `w` write (erases), `a` append, `r+` both. Always use `with` (auto-closes file).

```python
with open("notes.txt", "w") as f:
    f.write("Line 1\nLine 2\n")
with open("notes.txt", "a") as f:
    f.write("Line 3\n")
with open("notes.txt") as f:
    print(f.read())
with open("notes.txt") as f:
    for line in f: print(line.strip())

import json
data = {"name": "Asha", "skills": ["Python", "ML"]}
with open("data.json", "w") as f: json.dump(data, f, indent=2)
with open("data.json") as f: print(json.load(f))
```

#### 🔍 How it works
- `open(file, "w")` write (erases), `"a"` append, `"r"` read. **`with`** closes the file automatically even if an error occurs.
- `json.dump(obj, f)` writes Python dict/list as JSON text; `json.load(f)` reads it back.
#### 💼 Where it's used
Reading/writing logs, configs, exports. **JSON is the language of web APIs**, so this is how Python talks to services.

## 8.3 OOP — Classes & Objects
A **class** is a blueprint; an **object** is a thing made from it. Four pillars: encapsulation, inheritance, polymorphism, abstraction.

```python
class Student:
    school = "ABC School"                     # class attribute

    def __init__(self, name, marks):     # constructor
        self.name = name                 # instance attributes
        self.marks = marks

    def grade(self):
        return "Pass" if self.marks >= 40 else "Fail"

    def __str__(self):
        return f"{self.name} ({self.marks})"

s1 = Student("Riya", 85)
print(s1, s1.grade(), s1.school)

class Topper(Student):                   # inheritance
    def grade(self):                     # polymorphism (override)
        return "Distinction 🏆"
print(Topper("Sam", 99).grade())
```

#### 🔍 How it works
- `class Student:` = blueprint; `Student("Riya", 85)` = an **object** (instance). `__init__` runs automatically on creation; `self` = "this object".
- `school = ...` (class attribute) is shared by all objects; `self.name` (instance attribute) is per object.
- `class Topper(Student)` **inherits** everything and **overrides** `grade()` (polymorphism).
- `__str__` controls what `print(obj)` shows.
#### 💼 Where it's used
Every ML library is OOP: `LinearRegression()` is an object with methods `fit`, `predict`, `score`. Large applications are organised as classes.

## 8.4 Comprehensions
Short way to build lists/sets/dicts.

```python
print([n**2 for n in range(6)])
print([n for n in range(20) if n % 3 == 0])
print(["even" if n % 2 == 0 else "odd" for n in range(4)])
print({w: len(w) for w in ["apple", "fig"]})
print({c for c in "hello"})
print([[i * j for j in range(1, 4)] for i in range(1, 4)])   # nested
```

#### 🔍 How it works
- `[expr for item in iterable if condition]` = a loop + filter + append in one line: `[n**2 for n in range(6)]`.
- Same idea for dict `{k: v ...}` and set `{x ...}`. Nested comprehension builds a 2-D table (multiplication table).
#### 💼 Where it's used
Fast, readable data transformations: cleaning a list of strings, building feature lists, creating dictionaries from columns.
